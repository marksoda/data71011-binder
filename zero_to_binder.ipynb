### 2. Generate a random value
The tutorial uses an unseeded random draw. Here the seed is fixed so a fresh run also reproduces this number.

```python
rng = np.random.default_rng(42)
print("Seeded normal draw:", rng.standard_normal())
```

### 3. Read the Gapminder Oceania data
Rows are countries; columns contain GDP per capita for different years. Source: Software Carpentry, linked from the tutorial.

```python
data = pd.read_csv("gapminder.csv", index_col="country")
data.columns = data.columns.str.removeprefix("gdpPercap_").astype(int)
data
```

### 4. Plot Australia

```python
ax = data.loc["Australia"].plot(marker="o", figsize=(8, 4))
ax.set(title="Australia: GDP per capita", xlabel="Year", ylabel="GDP per capita (dataset units)")
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```
